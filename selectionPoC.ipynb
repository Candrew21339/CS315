```python
import random
import datetime
```

```python
def selection(a,print_flag):
  n = len(a)
  for i in range(n-1):
    smallest = i
    for j in range(i+1, n):
      if a[smallest] > a[j]:
        smallest = j
    a[smallest], a[i] = a[i], a[smallest]


# for your convenience, the psuedocode from class is provided, and note that you don't really need line 1, you can handle that in the outer for loop statement
#n = length[A]
#for i=1 to n-1:
#   smallest = i
#   for j=i+1 to n:
#      if A[smallest] > A[j]
#         then smallest = j
#   Swap A[i] with A[smallest]
```

```python

def selection_timed(a):
  global time_results
  print("n=" + str(n))
  start_time = datetime.datetime.now()
  selection(a,True)
  end_time = datetime.datetime.now()
  time_elapsed = (end_time - start_time).total_seconds()
  print(time_elapsed)

  #for ratio tests
  time_results.append(time_elapsed)
```

```python
#random
time_results = []

n=1000
a = list(range(n))
random.shuffle(a)
selection_timed(a)

n=2000
a = list(range(n))
random.shuffle(a)
selection_timed(a)

n=4000
a = list(range(n))
random.shuffle(a)
selection_timed(a)

n=8000
a = list(range(n))
random.shuffle(a)
selection_timed(a)

n=16000
a = list(range(n))
random.shuffle(a)
selection_timed(a)



N_values = [1000, 2000, 4000, 8000, 16000]


print("ratio test:")

for i in range(1, len(time_results)):
    ratio = time_results[i] / time_results[i-1]
    n_ratio = (N_values[i] / N_values[i-1])
    print(f"T({2**(i) * 1000}) / T({2**(i-1) * 1000}) = {ratio:.4f}")

print("Average ratio")

# Take the sum of all ratios, divide it by the number of ratios.
print(sum(time_results[i] / time_results[i-1] for i in range(1, len(time_results))) / (len(time_results) - 1))
```
