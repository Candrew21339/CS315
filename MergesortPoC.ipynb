```python
import random
import datetime
```

```python
# Merge provided as a convenience

def Merge(A,p,q,r):
    nL = q-p+1
    nR = r-q
    Left = []
    Right = []

    for i in range(nL):
        Left.append(A[p+i])
    for j in range(nR):
        Right.append(A[q+j+1])

    i=0
    j=0
    k=p
    while i<nL and j<nR:
        if (Left[i]<=Right[j]):
            A[k]=Left[i]
            i+=1
        else:
            A[k]=Right[j]
            j+=1
        k+=1

    while i< nL:
        A[k]=Left[i]
        i+=1
        k+=1
    while j< nR:
        A[k]=Right[j]
        j+=1
        k+=1

    return
```

```python
def Mergesort(A,p,r,print_flag):
    if p >= r:
        return
    q = (p+r)//2

    Mergesort(A,p,q,print_flag)
    Mergesort(A,q+1,r,print_flag)
    Merge(A,p,q,r)

    if print_flag:
      print(A)
    return

```
