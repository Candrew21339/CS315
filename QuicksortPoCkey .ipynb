```python
import random
import datetime
```

```python
# Partition provided as a convenience

def Partition(A,p,r):
    x = A[r]
    i = p - 1
    for j in range(p,r):
        if ( A[j] <= x ):
            i = i + 1
            #swap A[i] <--> A[j]
            temp=A[i]
            A[i]=A[j]
            A[j]=temp
    #swap A[i+1] <--> A[r]
    temp=A[i+1]
    A[i+1]=A[r]
    A[r]=temp

    return i + 1
```

```python
def Quicksort(A,p,r,print_flag):
    if p < r:
        q  = Partition(A , p , r)
        if print_flag:
            print(A)
        Quicksort(A , p , q-1,print_flag)
        Quicksort(A , q+1 , r,print_flag)
```

```python
def quicksort_timed(A,p,r,print_flag):
  global time_results
  print("n=" + str(N))
  start_time = datetime.datetime.now()
  Quicksort(A,p,r,print_flag)
  end_time = datetime.datetime.now()
  time_elapsed = (end_time - start_time).total_seconds()
  print(time_elapsed)

  #for ratio tests
  time_results.append(time_elapsed)
```

```python
#random
time_results = []


N=1000
A = list(range(N))
random.shuffle(A)
quicksort_timed(A,0,N-1,False)

N=2000
A = list(range(N))
random.shuffle(A)
quicksort_timed(A,0,N-1,False)

N=4000
A = list(range(N))
random.shuffle(A)
quicksort_timed(A,0,N-1,False)

N=8000
A = list(range(N))
random.shuffle(A)
quicksort_timed(A,0,N-1,False)

N=16000
A = list(range(N))
random.shuffle(A)
quicksort_timed(A,0,N-1,False)

N=32000
A = list(range(N))
random.shuffle(A)
quicksort_timed(A,0,N-1,False)

N=64000
A = list(range(N))
random.shuffle(A)
quicksort_timed(A,0,N-1,False)

print(time_results)



#Ratio tests.

N_values = [1000, 2000, 4000, 8000, 16000, 32000, 64000]
print("ratio test:")

for i in range(1, len(time_results)):
    ratio = time_results[i] / time_results[i-1]
    n_ratio = (N_values[i] / N_values[i-1])
    print(f"T({2**(i) * 1000}) / T({2**(i-1) * 1000}) = {ratio:.4f}")

print("Average ratio")

# Take the sum of all ratios, divide it by the number of ratios.
print(sum(time_results[i] / time_results[i-1] for i in range(1, len(time_results))) / (len(time_results) - 1))

```
