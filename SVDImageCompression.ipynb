```python
# @title
A=imread('Dog.jpg')
X=np.mean(A,-1);
```

```python
# @title
img=plt.imshow(A)
img.set_cmap('gray')
plt.axis('off')
plt.show()
```

```python
# @title
U, S, VT = np.linalg.svd(A,full_matrices=False)
S=np.diag(S)
```

```python
# @title
k=5
Aapprox=U[:,:k] @ S[0:k,:k] @ VT[:k,:]
plt.figure(1)
img=plt.imshow(Aapprox)
img.set_cmap('gray')
plt.axis('off')
plt.title('k = '+str(k))
plt.show()
```
