```python
# Template for testing Insertion-Sort on small lists, outputing each intermediate step
```

```python
import random
import datetime
```

```python
def insertion(A, print_flag):
  n = len(A)
  for i in range(1, n):
    key = A[i]
    j = i - 1
    while j >= 0 and A[j] > key:
      A[j + 1] = A[j]
      j = j - 1
      A[j + 1] = key
```

```python
def insertion_timed(a):
  global time_results
  print("n=" + str(n))
  start_time = datetime.datetime.now()
  insertion(a, False)
  end_time = datetime.datetime.now()
  time_elapsed = (end_time - start_time).total_seconds()
  print(time_elapsed)

  #for ratio tests
  time_results.append(time_elapsed)
```

```python
#random
time_results = []

n=1000
a = list(range(n))
random.shuffle(a)
insertion_timed(a)

n=2000
a = list(range(n))
random.shuffle(a)
insertion_timed(a)

n=4000
a = list(range(n))
random.shuffle(a)
insertion_timed(a)

n=8000
a = list(range(n))
random.shuffle(a)
insertion_timed(a)

n=16000
a = list(range(n))
random.shuffle(a)
insertion_timed(a)

N_values = [1000, 2000, 4000, 8000, 16000]


print("ratio test:")

for i in range(1, len(time_results)):
    ratio = time_results[i] / time_results[i-1]
    n_ratio = (N_values[i] / N_values[i-1])
    print(f"T({2**(i) * 1000}) / T({2**(i-1) * 1000}) = {ratio:.4f}")

print("Average ratio")

# Take the sum of all ratios, divide it by the number of ratios.
print(sum(time_results[i] / time_results[i-1] for i in range(1, len(time_results))) / (len(time_results) - 1))
```
