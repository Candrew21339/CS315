```python
import random
import datetime

def quickselect(A, i):
    # Select ith smallest element (1-indexed)
    if len(A) == 1:
        return A[0]
    pivot = random.choice(A)
    lows = [x for x in A if x < pivot]
    highs = [x for x in A if x > pivot]
    pivots = [x for x in A if x == pivot]
    if i <= len(lows):
        return quickselect(lows, i)
    elif i <= len(lows) + len(pivots):
        return pivot
    else:
        return quickselect(highs, i - len(lows) - len(pivots))

def quickselect_timed(a, i):
    global time_results
    n = len(a)
    print("n=" + str(n))
    start_time = datetime.datetime.now()
    result = quickselect(a.copy(), i)
    end_time = datetime.datetime.now()
    time_elapsed = (end_time - start_time).total_seconds()
    print(f"Time: {time_elapsed}, Result: {result}")
    time_results.append(time_elapsed)
```

```python
# Test on random data
print("=== RANDOM DATA ===")
time_results = []
N_values = [1000, 2000, 4000, 8000, 16000]

for n in N_values:
    a = list(range(n))
    random.shuffle(a)
    quickselect_timed(a, n // 2)

print("\nRatio test (O(n) expected):")
for n, t in zip(N_values, time_results):
    ratio = t / n
    print(f"n={n:5d}: T(n)/n = {ratio:.9f}")
```

```python
# Test on sorted data
print("\n=== SORTED DATA ===")
time_results = []
N_values = [1000, 2000, 4000, 8000, 16000]

for n in N_values:
    a = list(range(n))
    quickselect_timed(a, n // 2)

print("\nRatio test (O(n) expected):")
for n, t in zip(N_values, time_results):
    ratio = t / n
    print(f"n={n:5d}: T(n)/n = {ratio:.9f}")
```

```python
# Test on reverse sorted data
print("\n=== REVERSE SORTED DATA ===")
time_results = []
N_values = [1000, 2000, 4000, 8000, 16000]

for n in N_values:
    a = list(range(n, 0, -1))
    quickselect_timed(a, n // 2)

print("\nRatio test (O(n) expected):")
for n, t in zip(N_values, time_results):
    ratio = t / n
    print(f"n={n:5d}: T(n)/n = {ratio:.9f}")
```

```python
# Test on all same elements
print("\n=== ALL SAME ELEMENTS ===")
time_results = []
N_values = [1000, 2000, 4000, 8000, 16000]

for n in N_values:
    a = [5] * n
    quickselect_timed(a, n // 2)

print("\nRatio test (O(n) expected):")
for n, t in zip(N_values, time_results):
    ratio = t / n
    print(f"n={n:5d}: T(n)/n = {ratio:.9f}")
```
